<h3>자동 미분 (Automatic Differentiation)</h3>
손실 함수를 최소화하기 위해 각 가중치를 어떻게 변경해야하는지 자동으로 계산
<br><br>

<h4>1. Gradient (기울기)</h4>
미분값(편미분값), Weight의 변화에 따른 Loss 변화가 있는데 Loss가 감소하는 방향으로 Weight를 조절해야 함
<br><br>

<h4>2. Backpropagation (역전파)</h4>
Gradient를 계산하는 알고리즘
<br><br>

순전파 (Forward Propagation)
<ul>
    <li>입력 값을 넣어서 결과를 계산함 <br> ex) x → Neural Network → y_결과,  Loss = y_실제값 - y_결과 (손실 계산)
    </li>
</ul>

역전파 (Backpropagation)
<ul>
    <li>손실이 계산되면 출력층 → 입력층 방향으로 거꾸로 이동하면서 각 Weight의 Gradient를 계산함</li>
</ul>

<h4>Tensorflow의 자동 미분</h4>
<ul>
    <li>
        신경망에서 가중치가 수백만 개일 수 있음. 이런 네트워크의 미분을 사람이 계산하는 것을 불가
    </li>
    <li>
        Tensorflow는 계산 값만 저장하는 것이 아닌 GradientTape가 계산되는 과정도 모두 기록<br>(ex : x → y, y → z, z → loss)
    </li>
    <li>
        x - x 2 -> y - + 3 -> z - ^ 2 -> loss (이러한 구조를 계산 그래프라고 함 (Computational Graph)
    </li>
    <li>
        이 계산 그래프를 저장했다가 gradient()가 기록된 경로를 역방향으로 추적하여 미분값을 계산해줌
    </li>
</ul>

<h4>3. GradientType</h4>
자동으로 미분을 수행하는 객체

```python
import tensorflow as tf

x = tf.Variable(3.0)

with tf.GradientTape() as tape: # GradientTape 생성
    y = x ** 2                  # 계산 기록

grad = tape.gradient(y, x)      # 계산 수행 (y' = 2x, x = 3.0)

print(grad.numpy())
```

```python
# 여러 변수의 Gradient
x = tf.Variable(2.0)
y = tf.Variable(3.0)

with tf.GradientTape() as tape:
    z = x*x + y*y

grads = tape.gradient(z, [x, y]) # (∂z / ∂x = 2x = 4), (∂z / ∂y = 2y = 6)

print(grads)
```
