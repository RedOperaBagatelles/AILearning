### 신경망
인간의 뇌 신경망(Neuron)을 모방하여 만든 모신러닝 모델

#### 1. Perceptron (퍼셉트론)
가장 단순한 형태의 인공 신경망

<ul>
    <li>입력 값에 가중치를 곱한 후 모두 더하고, 편향(Bias)을 더해서 결과를 출력</li>
</ul>

$$ y = f(w_1x_1 + w2x2 + ... + ) $$

<ul>
    <li>x : 입력 값, w : 가중치, b = 편향, f : 활성 함수</li>
</ul>

Perceptron의 한계 : XOR 문제를 해결할 수 없음 (하나의 직선으로 데이터를 구분할 수 없기 때문)

#### 2. Multi Layer Perceptron (MLP)
퍼셉트론을 여러 층으로 쌓은 구조

$$ 입력층 → 은닉층 → 은닉층 → 출력층$$

<ul>눈, 코, 입 → 얼굴 → 사람 처럼 점점 고차원 특징을 추출할 수 있음</ul>

#### 3. Forward Propagation (순전파)
입력값이 출력까지 전달되는 과정 (딥러닝의 예측 단계)

<ul>
    <li>과정 : 입력(x) → 가중치 계산 ($z = $W$x + $b) → 활성화 함수 ($a = $f($z)) → 다음층 전달</li>
</ul>


#### 4. Backpropagation (역전파)
신경망이 학습하는 핵심 알고리즘 (예측 결과와 실제 결과의 차이[오차]를 계산하고 가중치를 수정)

<ul>
    <li>과정 : 예측 → 오차 계산 → 오차를 뒤에서 앞으로 전파 → 가중치 수정</li>
</ul>

$$ W = W - η{∂Loss \over  ∂W} $$
$$ 오차\;수정\;계산식 \; [η : 학습률 \; (Learning \; Rate)] $$
