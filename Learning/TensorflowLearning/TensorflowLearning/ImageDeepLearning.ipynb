### 이미지 딥러닝
이미지 딥러닝은 컴퓨터가 사진이나 그림 속의 특징응 스스로 학습하여 분류하거나 인식하는 기술
대표적으로 CNN(Convolutional Neural Network)를 사용함

#### 1. CNN (Convolutional Neural Network)
이미지 처리에 특화된 딥러닝 모델

<ul>
    <li>이미지의 픽셀 수가 많아져 학습해야 할 데이터가 너무 많아지면, 이미지 특징만 추출하여 효과적으로 학습</li>
    <li>ex) 숫자 이미지 -> 0 ~ 9 분류, 얼굴 인식, 자율주행 차량의 객체 인식</li>
</ul>

#### 2. 합성곱 (Convolution)
이미지 특징을 추출하는 과정

<ul>
    <li>작은 필터 (Kernel)를 이미지 위에서 이동시키며 계산</li>
    <li>이미지 데이터 : 1 1 1 0 1 0 0 0 1, 필터 : 1 0 1 0 -> 이 필터를 이미지 데이터 위에 이동 시키면서 곱한 값을 더해 새로운 특징 맵을 생성</li>
    <li>추출 가능한 특징 : 초기층 [선, 모서리], 중간층 [눈, 코, 입], 깊은층 [얼굴, 차, 동물]</li>
</ul>

```python
# 이미지 특징 추출 예시 코드
from tensorflow.keras.layers import Conv2D

Conv2D(filters=32, kernel_size=(3,3), activation='relu')
```

주요 파라미터

<ul>
    <li> filters : 필터 개수, kernel_size : 필터 크기, strides : 이동 간격, padding : 가장자리 처리, activation : 활성화 함수</li>
</ul>

#### 4. 폴링 (Pooling)
Convolution 후 이미지 크기를 줄이는 과정

<ul>
    <li>목적 : 계산량 감소, 과적합 방지, 중요한 특징 유지</li>
    <li>Max Pooling : 가장 큰 값을 선택하는 방법 (가장 많이 사용됨)</li>
</ul>

```python
# Max Pooling 예시
from tensorflow.keras.layers import MaxPooling2D

MaxPooling2D(pool_size=(2,2))
```

#### 4. CNN 기본 구조
일반적인 CNN 구조

<ul>
    <li>입력 이미지 -> Convolution -> ReLU -> Pooling -> Convolution -> Pooling -> Flatten -> Dense -> Output</li>
    <li>Flatten : 2차원 이미지를 1차원 백터로 변경</li>
</ul>

#### 5. 이미지 분류 (Image Classification)
이미지를 특정 클래스에 분류하는 역할


```python
# CNN 분류 모델 예제
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense

model = Sequential([
    Input(shape=(28, 28, 1)),
    
    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.summary()
```
