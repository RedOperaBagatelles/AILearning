#### 1. Sequential
<ul>
    <li>레이어를 순서대로 쌓는 가장 단순한 모델</li>
    <li>입력 → 은닉층1 → 은닉층2 → 출력층</li>
</ul>

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()
```

```python
# 또는
model = Sequential([Dense(64, input_shape=(10, )),
                    Dense(32),
                    Dense(1)
                   ])
```

##### 위 구조의 이해
<ul>
    <li>입력 데이터 (10개) → Dense (64) → Dense (32) → Dense(1)</li>
    <li>데이터가 왼쪽에서 오른쪽으로 순차적으로 이동</li>
</ul>

##### 특징
장점
<ul>
    <li>사용법이 간단</li>
    <li>초보자가 배우기 쉬움</li>
    <li>대부분의 기본적인 신경망 구현 가능</li>
</ul>

단점
<ul>
    <li>복잡한 구조 구현이 어려움</li>
    <li>여러 입력/출력 모델 불가능</li>
</ul>

#### 2. Dense Layer

<ul>
    <li>가장 기본적인 완전 연결층(Fully Connected Layer)</li>
    <li>모든 입력 노드가 모든 출력 노드에 연결된 모습</li>
    <li>Dense(N) == N개의 출력 노드를 가짐</li>
    <li>입력층 4개, 출력층 3개인 경우<br>
        x1 ─┬── y1<br>
         &nbsp; &nbsp; &nbsp;  &nbsp;├── y2<br>
         &nbsp; &nbsp; &nbsp;  &nbsp;└── y3<br><br>
        x2 ─┬── y1<br>
         &nbsp; &nbsp; &nbsp;  &nbsp;├── y2<br>
         &nbsp; &nbsp; &nbsp;  &nbsp;└── y3</li>
    <li>위 그림과 같이 형성</li>
</ul>

<ul>
    <li>Dense Layer에서 사용하는 수식 $$y = Wx + b $$</li>
    <li>y == 출력값, W == 가중치, x == 입력값, b == 편향 </li>
    <li>각각의 모든 입력 노드를 통해 데이터를 입력 받고 가중치를 각각 곱한 후 합치고 편향 값을 더함</li>
</ul>

```python
# 실제 코드
Dense(units=64, activation='relu') # units == 노드 수, activation == 활성화 함수
```

#### 3. Activation Function (활성 함수)
Dense Layer의 계산 결과를 비선형성을 추가하는 함수<br><br>

<ul>
    <li>활성 함수 없이 Dense → Dense → Dense 방식으로 여러 개 쌓아도 하나의 선형 모델과 같아짐</li>
    <li>단순한 패턴이 아닌 복잡한 형태로 학습하기 위해서 활성 함수를 사용함</li>
</ul>
<br>

동작 과정<br>
입력 → Dense 계산 → Wx + b → 활성 함수 → 출력

대표적인 활성 함수
1. ReLU (가장 많이 사용)
$$ f(x) = max(0, x) $$

<ul>
    <li>음수는 0, 양수는 그대로 통과</li>
    <li>ex) [x = -5 → 0], [x = -1 → 0], [x =  3 → 3], [x = 10 → 10]</li>
    <li>장점 : 계산이 빠름, 학습 속도 우수, 현재 가장 많이 사</li>
</ul>
<br>

2. Sigmoid
$$ 𝜎(x) = {1 \over 1 + e^{-x}} $$

<ul>
    <li>출력을 0 ~ 1 사이로 변환</li>
    <li>ex) [x = -10 → 0.0000], [x = 0 → 0.5000], [x = 10  → 0.9999]</li>
    <li>사용 분야 : 이진 분류 (Yes or No)</li>
</ul>
<br>

3. SoftMax

<ul>
    <li>여러 클래스 중 하나를 선택해야 하는 경우 (ex : 고양이, 강아지, 토끼, MNIST)</li>
    <li>출력 예) [2.0, 1.0, 0.1] → [0.65, 0.24, 0.11] (확률의 합은 항상 1)</li>
</ul>

```python
# 3가지를 활용한 예제
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 입력 10개 → Dense (64) → ReLU → Dense (32) → ReLU → Dense (1) → Sigmoid → 0 ~ 1
model = Sequential([Dense(64, activation='relu', input_shape=(10,)),
                    Dense(32, activation='relu'),
                    Dense(1,  activation='sigmoid')
                   ])
```
