### 모델 학습 (Model Training)
모델 학습 과정 : compile() [학습 방법 설정] → fit() [모델 학습] → evaluate() [모델 성능 평가] → predict() [예측 수행]

#### 1. compile()
모델이 어떻게 학습하는지를 설정하는 단계

```python
model.compile(optimizer='adam', loss='mse', metris=['mae'])
```

optimizer 주요 옵션

<ul>
    <li>SGD : 가장 기본적인 경사하강법</li>
    <li>Adam : 가장 많이 사용</li>
    <li>RMSprop : 시계열 데이터에 자주 사용</li>
    <li>Adagrad : 학습률을 자동 조정</li>
</ul>

loss 주요 옵션

<ul>
    <li>회귀(Regression) : mse [Mean Squared Error : 오차가 클수록 큰 패널티]</li>
    <li>이진 분류 : binary_crossentropy</li>
    <li>다중 분류 : categorical_crossentropy</li>
</ul>

matrix
학습 중 확인할 성능 지표

<ul>
    MAE(Mean Absolute Error) : |실제값 - 예측값|의 평균
</ul>

#### 2. fit()
실제 데이터를 사용하여 모델을 학습

```python
history = model.fit(x_train, y_train, epochs=100, batch_size=32, validation_split=0.2)
```

epochs

<ul>
    <li>전체 데이터를 몇 번 학습할지 설정</li>
    <li>ex : epoches=100 (100번 학습)</li>
</ul>

batch_size

<ul>
    <li>한 번에 학습할 데이터 수</li>
    <li>ex : batch_size=10 (데이터 100개일 때, 10개씩 10번 학습)</li>
</ul>

validation_split

<ul>
    <li>검증 데이터 비율</li>
    <li>ex : validation_split=0.2 (전체 데이터의 20%를 검증용으로 사용)</li>
</ul>

loss가 감소하면 학습이 잘 진행된 것

### 3. evaluate()
학습이 끝난 후 테스트 데이터로 성능 측정

```python
loss, mae = model.evaluate(x_test, y_test)
```

<ul>
    <li>loss : 오차, mae = 평균 절대 오차</li>
</ul>

#### 4. predict()
학습된 모델을 사용하여 새로운 데이터 예측

```python
result = model.prdict([[10]]) # [결과(예측) : 20.13]
```

```python
# 전체 코드

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 모델 생성
model = Sequential([Dense(8, activation='relu', input_shape=(1,)), Dense(1)])

#학습 설정
model.compile(optimizer='adam', loss='mse', matrix=['mae'])

# 학습
model.fit(x_train, y_train, epochs=100)

#평가
model.evaluate(x_test, y_test)

#예측
result = model.predict([[10]])
print(result)
```
