### 자연어 처리 (NLP)
사람이 사용하는 언어(텍스트)를 컴퓨터가 이해하고 처리할 수 있도록 하는 인공지능 분야

<ul>
    <li>예시 : 챗봇, 번역기, 문서 요약, 감정 분석, 문서 분류, 검색 엔진, 음성 인식</li>
    <li>처리 순서 : 텍스트 입력 → Tokenization (문장을 토큰으로 분리) → Embedding(토큰을 의미 있는 숫자 벡터로 변환) → RNN / LSTM (순서와 문맥을 학습) → 출력 (번역, 감정 분석 문장 생성, 분류 등)</li>
</ul>

#### 1. Tokenization (토큰화) 
가장 먼저 수행되는 과정

<ul>
    <li>문장을 작은 단위(Token)로 나누는 과정</li>
    <li>컴퓨터는 문장 그래도 이해하지 못하기 때문에 문장을 하나의 단어 또는 "안녕", "하", "세요"처럼 여러 토큰으로 나누어 처리</li>
    <li>ex : 나는 오늘 학교에 갔다. → ["나는", "오늘", "학교에", "갔다"]</li>
    <li>ex : playing → ["play", "##ing"] (BERT 방식)</li>
</ul>

Tokenization 종류
<ul>
    <li>Word Tokenization : 단어 기준 (I love Python → ["I", "love", "Python"])</li>
    <li>Character Tokenization : 글자 기준 (Hello → ["H", "e", "l", "l", "o"])</li>
    <li>Subword Tokenization : 가장 많이 사용 (unbelievable → ["un", "believ", "able"] 또는 ["un", "##believ", "##able"])</li>
    <ul>
        <li>장점 : 새로운 단어도 처리 가능, Vocabulary 크기 감소</li>
        <li>대표 기법 : BPE(Byte Pair Encoding), WordPiece, SentencePiece</li>
    </ul>
</ul>

#### 2. Embedding (임베딩)
Token을 숫자로 변환하는 과정

<ul>
    <li>컴퓨터는 숫자만 계산할 수 있기 때문</li>
    <li>사과를 단순히 "사과 = 1, 바나나 = 2, 포도 = 3"처럼 단순 번호로 지정하면 의미를 반영하지 못함</li>
    <li>사과 → [0.13, 0.82, -0.55, 0.24, ...] 처럼 백터를 이용하여 단어를 여러 개의 숫자로 표현</li>
    <li>왕 → [0.91, 0.34, ...]은 여왕 → [0.89. 0.37, ...]와 거의 비슷한 위치에 존재, 자동차는 전혀 다른 위치에 존재</li>
    <li>유명한 Embedding : Word2Vec, GloVe, FastText, BERT Embedding</li>
</ul>

#### 3. RNN (Recurrent Neural Network)
순서가 있는 데이터를 처리하기 위한 신경망

<ul>
    <li>일반적인 싱경망은 A → B → C 각 입력 독립적으로 처리</li>
    <li>"먹었다 나는 밥을."는 틀린 문장이고 "나는 밥을 먹었다."는 맞는 것처럼 문장은 순서가 매우 중요</li>
    <li>RNN은 이전 정보를 기억하면서 다음 단어를 처리 (ex : 오늘 → 오늘 날씨 → 오늘 날씨 좋다)</li>
    <li>구조 : 단어 1 → Hidden State → 단어 2 → Hidden State → 단어 3 (Hidden State가 기억 역할을 함)</li>
    <li>장점 : 순서 정보 활용, 문자 처리 가능, 기계열 데이터 처리 가능</li>
    <li>단점 : 긴 문장을 잘 기억하지 못하는 문제 (앞부분 내용을 잊어버리는 장기 의존성 [Long-term Dependency] 문제)</li>
</ul>

#### 4. LSTM (Long Short-Term Memory)
LSTM은 RNN을 개선한 모델

<ul>
    <li>긴 문장도 잘 기억할 수 있게 설계었음</li>
    <li>LSTM은 마지막 단어를 이해할 때 기존의 정보도 계속 기억해야 함</li>
    <li>구조 : 세가지 게이트를 사용</li>
    <li>입력 → Forget Gate (잊을 정보) → Input Gate (새로운 정보 저장) → Output Gate (출력할 정보) → 다음 상태</li>
    <li>장점 : 긴 문장 처리 가능, 번역 성능 향상, 감정 분석에 적합, 음성 인식에도 많이 사용</li>
</ul>

① Forget Gate
<ul>
    <li>필요 없는 정보를 제거</li>
    <li>오늘은 → 날씨는 → ... (이미 필요 없는 정보는 삭제)</li>
</ul>

② Input Gate
<ul>
    <li>새로운 정보를 저장</li>
    <li>새로운 단어 입력 → 중요하면 기억</li>
</ul>

③ Output Gate
<ul>
    <li>현재 필요한 정보만 출력</li>
    <li>기억 → 현재 출력</li>
</ul>

#### 5. GRU (Gated Recurrent Unit)
LSTM을 간소화한 순환 신경망(RNN)

<ul>
    <li>기존 LSTM은 긴 문장을 기억하기 위해 3개의 게이트를 사용하지만 구조가 복잡하고 계산량이 많음</li>
    <li>구조 : 입력 Reset Gate (이전 정보 얼마나 사용할지) → Update Gate (새 정보를 얼마나 저장할지) → 출력</li>
</ul>

① Reset Gate
<ul>
    <li>현재 문맥과 관련 없는 이전 정보는 일부 제거 (정확히는 정보를 지우는 것이 아닌 사용할 가중치를 줄임)</li>
    <li>"나는 어제 부산에 갔다. 오늘은 서울에 있다."에서 현재 위치는 서울을 가중치를 높게하고 과거 위치인 부산의 가중치를 낮게 설정</li>
</ul>

② Update Gate

<ul>
    <li>기존 기억과 새로운 정보를 얼마나 반영할지 결정</li>
    <li>기존 기억 + 새로운 정보 → 새로운 Hidden State</li>
    <li>장점 : LSTM보다 계산이 빠름, 메모리 사용량이 적음, 긴 문장도 어느 정도 처리 가능, 구현이 간단 \</li>
</ul>

#### 6. Attention (어텐션)
문장에서 중요한 단어에 집중하는 메커니즘

<ul>
    <li>필요 순간에 문장의 모든 단어를 다시 참고하여 중요한 정보를 선택</li>
    <li>ex : 문장 (철수는 사과를 먹고 학교에 갔다.), 질문 (누가 학교에 갔는가?), Attenstion (철수에 높은 중요도를 부여)</li>
    <li>장점 : 긴 문장 처리 가능, 번역 성능 향상, 문맥 이해 능력 향상</li>
</ul>

#### 7. Transformer
Attention을 기반으로 만든 모델로 RNN/LSTM처럼 순서대로 읽기 않고 문장을 한 번에 입력받아 병렬 처리함

<ul>
    <li>학습 속도가 매우 빠르고 긴 문장도 효과적으로 처리</li>
    <li>구조 : 입력 → Embedding → Positional Encoding (단어 순서 정보 추가) → Encoder → Decoder → 출력</li>
    <li>장점 : 병렬 처리 가능, 긴 문장 처리 우수, Attension 사용, 현재 NLP의 표준 모델</li>
</ul>

Positional Encoding

<ul>
    <li>Transformer는 문장을 한 번에 처리하므로 순서를 알 수 없음</li>
    <li>위치 정보를 함께 추가함</li>
    <li>ex : 나는(위치 1) 오늘(위치 2) 학교(위치3)</li>
</ul>

Encoder

<ul>
    <li>문장의 의미를 이해</li>
    <li>입력 : 나는 학교에 갔다 → 문장의 특징 추출</li>
</ul>

Decoder

<ul>
    <li>Encoder의 결과를 이용해 새로운 문장을 생성</li>
    <li>예시 : 영어 → 한국어, 질문 → 답변 생성</li>
</ul>

#### 8. BERT (Bidirectional Encoder Representations from Trnsformers)
Transformer의 Encoder만 사용하는 모델

<ul>
    <li>양방향(Bidirectional)으로 문장을 이해함</li>
    <li>앞 단어 → 현재 단어 → 뒤 단어 를 모두 함께 고려</li>
    <li>ex : 은행에 갔다 → 은행이 금융기관인지, 강가인지 앞뒤 문맥을 함께 보고 판단</li>
</ul>

학습 방법

<ul>
    <li>문장의 일부를 가리고 모델은 그 부분을 예측함 (Masked Language Model[MLM]이라고 함)</li>
    <li>입력 : 나는 [MASK]를 먹었다. → 예측 : 사과</li>
    <li>활용 분야 : 문장 분류, 감정 분석, 질의 응답, 검색, 개체명 인식</li>
    <li>특징 : 문장 이해에 매우 강함, 생성보다는 이해 중심</li>
</ul>

#### 9. GPT (Generative Pre-trained Transformer)
GPT는 Transformer의 Decoder만 사용하는 생성형 언어 모델

<ul>
    <li>ex : 오늘 날씨가 → 모델 예측 : 좋네요. → 모델 예측 : 산책하기 좋은 입니다.</li>
    <li>학습 방법 : 항상 단어를 예측하여 학습</li>
    <li>활용 분야 : 챗봇, 문서 작성, 코드 생성, 번역, 요약, 질의 응답</li>
    <li>특징 : 자연스러운 문장 생성, 긴 대화 유지, 다양한 작업 수행 가능</li>
</ul>
