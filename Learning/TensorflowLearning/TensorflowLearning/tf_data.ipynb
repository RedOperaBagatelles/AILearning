### 데이터 처리 (tf.data)
데이터를 얼마나 효율적으로 공급하느냐에도 성능에 미칠 수 있음
tensorflow에서는 tf.data API를 이용하여 데이터를 읽고, 전처리하고, 학습용으로 공급하는 파이프라인을 구축

```python
import tensorflow as tf

data = [1, 2, 3, 4, 5]
dataset = tf.data.Dataset.from_tensor_slices(data) # from_tensor_slicessms 배열, 리스트, numpy 배열 등을 Dataset 객체로 반환

for item in dataset:
    print(item.numpy())
```

#### 1. Dataset
Dataset은 Tensorflow에서 데이터를 관리하는 기본 객체

특징
<ul>
    <li>데이터를 순차적으로 제공</li>
    <li>메모리 효율적</li>
    <li>대용량 데이터 처리 가능</li>
    <li>학습 파이프라인 구축 가</li>
</ul>

#### 2. Batch
딥러닝을 보통 데이터를 하나씩 학습하지 않고 여러 개를 묶어서 학습

```python
dataset = tf.data.Dataset.from_tensor_slices([1, 2, 3, 4, 5, 6])
dataset = dataset.batch(2)

for batch in dataset:
    print(batch.numpy())
```

장점

<ul>
    <li>GPU 활용률 증가</li>
    <li>학습 속도 향상</li>
    <li>안정적인 Gradient 계산</li>
</ul>

#### 3. Shuffle
데이터 순서를 무작워로 섞음 (데이터가 정렬되어 있을 경우 학습 편향이 생길 수 있으므로 섞어서 학습 효과를 높임)

```python
dataset = tf.data.Dataset.from_tensor_slices([1, 2, 3, 4, 5])
dataset = dataset.shuffle(5)

for batch in dataset:
    print(batch.numpy())
```

#### 4. Prefetch
모델이 학습하는 동안 다음 데이터를 미리 준비

```python
dataset = dataset.prefetch(tf.data.AUTOTUNE)
```

<ul>
    <li>CPU는 다음 데이터 준비, GPU는 현재 패치를 학습하여 동시 작업 수행</li>
</ul>

장점
<ul>
    <li>GPU 대기 시간 감소</li>
    <li>전체 학습 속도 향상</li>
</ul>

#### 5. Cache
처음 읽은 데이터를 메모리에 저장해 두고 재사용

```python
# 1회차 : 디스크 읽기 → 전처리 → 학습, 2회차 : 메모리 사용 → 학습
dataset = (tf.data.Dataset.from_tensor_slices(data).cache())
```

장점

<ul>
    <li>디스크 I/O 감소</li>
    <li>데이터 로딩 속도 향상</li>
    <li>Epoch 반복 시 성능 향상</li>
</ul>

주의

<ul>
    <li>데이터가 너무 크면 메모리가 부족할 수 있음</li>
</ul>

```python
dataset = (tf.data.Dataset.from_tensor_slices(data)
           .shuffle(1000)
           .batch(32)
           .cache()
           .prefetch(tf.data.AUTOTUNE)
          )
```
